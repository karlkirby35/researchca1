## 2: Evidence-Based Recommendations:

### 2.1 Workforce Planning - Retail Analytics Findings

The company should review its workforce planning across different stores to ensure that staffing levels are appropriate for customer demand. Task 1 identified staffhours as the most influential feature in the random forest model, accounting for 37.79 percent of feature importance. Footfall was the second most influential feature at 19.43 percent. Staff hours has a positive correlation of 0.576 with salesrevenue.

These findings suggest staffing levels and customer footfall are important variables that are associated with the retail sales performance. The company should adjust what employees are at different shops to account for busy periods, such as Christmas. One small shop might get less customers than a bigger shop during busy periods and busy times. This ensures customer service expectations are met and the standards are kept to an appropriate level.

Managment should monitor staffing costs, customer service and sales performance when introducting these changes to staff rosters. This ensures the adjustments made are working across the different stores.

### 2.2 Regional and Seasonal Performance - Retail Analytics Findings

The company should use its regional and seasonal sales results to make better decisions about stock levels, marketing campaigns and business planning. Task 1 identified that the North generated the highest total regional value at approx €191.89 million, while the South generated the lowest at approx €186.24 million. However, the South had the highest average sales revenue per record, at €155,202.32. This suggests that management has to consider both total revenue and average performance when comparing regions.

Seasonal analysis also showed that summer generated the highest total revenue, at approx €197.81 million, while spring recorded the lowest at approx €183.77 million. The company should use these trends when planning seasonal promotions. These could include management reviewing stock availability during the summer period and ensuring the store has sufficient products for that period.

By monitoring these patterns, the Commerical Director can make informed decisions regarding the regional and seasonal decisions.

### 2.3 Personalised Product Recommendations - Recommendation System Findings

The company should introduce personalised product recommendations to help customers discover products that match their purchasing interests. Task 2 identified comparing a popular based reccomendation system with an item based collab filtering system. The collaborative filrtering approach achieved a precision score of 17.20 percent compared with 4.60 percent for the popularity approach.

These results suggest that collaborative filtering was more effective at recommending relevant products to customers than just recommending the most popular ones. The company could also use this approach on their website by displaying recommended products based on customers previous purchases.

This could assisit customers in discovering additional products. Before this is implemented the company should test it using a selection of customers and taking down evidence of how its going through the process.

### 2.4 Customer Segmentation - Recommendation System Findings

The company should use customer segmentation which could help customer engagement and more marketing strategys. Task 2 divided customers into four groups based on their spending. Low, Medium, High and Top value. The results showed a clear difference in customer spending patterns, with the low group spending on average €5,229.95 and the top group spending on average €16,894.80.

This information could help the company create marketing campaigns that are more relevant to different customer groups. Top value customers could receieve more personalised offers and more loyalty rewards, while the lower group could be targeted with discounts and discount bundles.

By using customer spending patterns to guide marketing decisions, the company could improve customer retention and identify ways to increase the customer value. Management should monitor the spending habits during special events to use as evidence for comparing the different habits.

### 2.5 Product and Category Performance - Recommendation System Findings

The company should use product and category data to identify which areas of the product range generates the most revenue. Task 2 identifies that books generated the highest revenue at approx €2.34 million, followed by gaming which was at €2.28 million. Electronics generated the lowest revenue which was €1.45 million.

The commerical director should use these findings when reviewing product availability, promotional activity and future product planning. The company should investigate promotions for the high revenue products and review the performance of the lower revenue products to try increase the revenue of them.

This approach could help the company make better decisions about the product range and marketing investment. However, the revenue should be considered alongside factors such as profit and marketiung investment.
